```python
x = np.array([[1,2],[3,4]])
y = np.array([[5,6],[7,8]])

v = np.array([9,10])
w = np.array([11, 12])

print(x.shape)
print(y.shape)
print(v.shape)
print(w.shape)

# Matrix / vector product; both produce the rank 1 array [29 67]
print(x.dot(v))
print(np.dot(x, v))

print(x@v)
print(v@x)
```

```python
for i in range(len(x)):
    print(x[i])
    for j in range(len(y)):
        print(f" {y[j]} {x[i].dot(y[j])}")

# Matrix / matrix product; both produce the rank 2 array
# [[19 22]
#  [43 50]]
print(x.dot(y))
print(np.dot(x, y))

print(x @ y)
```

```python
import torch
torch.manual_seed(123)
```

```python
a = torch.tensor([[1, 2, 3], [4, 5, 6]])   # from Python lists
print(a.shape, a.ndim, a.dtype)
```

```python
print(torch.zeros(2, 3).shape)             # sizes as separate numbers...
print(torch.ones((2, 3)).shape)            # ...or as one tuple: same thing
print(torch.rand(2, 3))                    # uniform [0, 1)
print(torch.arange(6))                     # 0..5, handy because you can see where each number moves
```

```python
x = torch.arange(24).reshape(2, 3, 4)      # (batch, tokens, features)
print(x)
print(x.shape)
```

```python
print(x[0].shape)         # (3,4)  
print(x[0:1].shape)       # (3,4) ---- X
print(x[:, -1].shape)     # (2,4) 
print(x[:, :-1].shape)    # (2,2,4) 
print(x[..., 0].shape)    # (2,3) 
print(x[:, None].shape)   # (2,1,3,4) 
print(x[..., None].shape) # (2,3,4,1) 
```

```python
s = torch.zeros(2, 3, 3)
m = torch.triu(torch.ones(3, 3), diagonal=1).bool()

print(s)
print(m)
```

```python
s[:, m] = -1               # ':' takes dim 0, the 2-D mask covers dims 1 and 2
print(s)
```

```python
y = torch.rand(2, 4, 6, 8)
print(y[0, ..., 1].shape)      # (4, 6)
print(y[..., 0, :].shape)      # (2, 4, 8)   2nd-to-last dim indexed, last kept
```

```python
logits = torch.rand(1, 6, 50257, 1)

print(logits.shape) 
print(logits.squeeze().shape) 
```

```python
w = torch.tensor([10, 20])                # (2,)  one value per ROW
v = torch.tensor([1, 2, 3])
print(v[:, None] * w)                     # (3, 1) * (2,) -> (3, 2)
```

```python
t = torch.arange(12)
print(t.reshape(3, 4))
print(t.reshape(3, -1).shape)   # -1 -> 4
print(t.reshape(-1, 6).shape)   # -1 -> 2
```

```python
x = torch.arange(24).reshape(2, 3, 4)       # (b, tokens, d) with d = 4
print(x.shape)
print(*x.shape[:-1])
```

```python
x = torch.arange(6).reshape(2, 3)
print(x)
print(x.transpose(0, 1))     # rows become columns: numbers move
print(x.reshape(3, 2))       # same shape (3, 2), but numbers stay in reading order -> different tensor!
```
